```python
from PIL import Image, ImageFilter, ImageEnhance, ImageOps
import matplotlib.pyplot as plt

# Загрузка изображения
image = Image.open("jZU1hT8Kmy0.jpg").convert("RGB")  # Преобразуем в RGB, если вдруг в другом формате

# Шаг 1: Удаление шума (медианный фильтр)
denoised_image = image.filter(ImageFilter.MedianFilter(size=3))

# Шаг 2: Повышение резкости
sharpened_image = denoised_image.filter(ImageFilter.UnsharpMask(radius=2, percent=150, threshold=3))

# Шаг 3: Повышение контраста
enhancer = ImageEnhance.Contrast(sharpened_image)
contrast_enhanced = enhancer.enhance(1.5)

# Шаг 4: Инверсия цветов
inverted_image = ImageOps.invert(contrast_enhanced)

# Отображение результатов
plt.figure(figsize=(18, 6))

plt.subplot(1, 3, 1)
plt.imshow(image)
plt.title("Оригинал")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(contrast_enhanced)
plt.title("После улучшения")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(inverted_image)
plt.title("После инверсии цветов")
plt.axis("off")

plt.tight_layout()
plt.show()

# Сохранение изображения
inverted_image.save("improve_image.jpg")
```
